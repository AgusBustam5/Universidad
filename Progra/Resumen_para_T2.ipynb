### 1. Herencia
El concepto: Una clase "hija" copia automáticamente todo lo de una clase "padre" y le añade lo suyo.
El problema real que resuelve: Evitar copiar y pegar el mismo código en 50 archivos distintos. Si encuentras un error en el código duplicado, tendrías que arreglarlo 50 veces. Con herencia, lo arreglas una vez en el padre.
Caso práctico (E-commerce):
Tienes productos físicos y digitales. Ambos tienen nombre, precio y un método comprar().
Creas una clase base Producto. Luego creas ProductoFisico (que hereda de Producto y añade calcular el costo de envío) y ProductoDigital (que hereda de Producto y añade generar un link de descarga).

```python
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

class ProductoFisico(Producto):
    def calcular_envio(self):
        return self.precio * 0.1

class ProductoDigital(Producto):
    def generar_enlace(self):
        return f"Descarga_{self.nombre}.zip"

libro = ProductoFisico("Libro Python", 20)
curso = ProductoDigital("Curso OOP", 50)
```

### 2. Polimorfismo
El concepto: Tratar a objetos diferentes exactamente de la misma manera, confiando en que cada uno sabrá cómo comportarse.
El problema real que resuelve: Evitar los bloques gigantes de if/elif que preguntan "qué tipo de objeto eres" antes de actuar.
Caso práctico (Pasarelas de Pago):
En una tienda online, tienes pagos con Paypal, Tarjeta y Criptomonedas.
En lugar de escribir:
si pago es paypal: procesar_paypal() elif pago es tarjeta: procesar_tarjeta()
Creas tres clases distintas, todas con un método llamado pagar(). El sistema de compras simplemente toma el método elegido y ejecuta metodo_elegido.pagar(). Si mañana añades ApplePay, el sistema principal no cambia en absoluto; solo agregas una nueva clase.

```python
class PayPal:
    def pagar(self, monto):
        return f"Procesando ${monto} vía PayPal"

class Tarjeta:
    def pagar(self, monto):
        return f"Procesando ${monto} vía Tarjeta"

def procesar_compra(metodo_pago, monto):
    # La función ignora si es PayPal o Tarjeta, solo ejecuta el método
    print(metodo_pago.pagar(monto))

procesar_compra(PayPal(), 100)
procesar_compra(Tarjeta(), 50)
```

### 3. Multiherencia
El concepto: Una clase hereda de dos o más clases padre al mismo tiempo.
El problema real que resuelve: Necesitas darle capacidades específicas a un objeto sin obligarlo a encajar en una sola rama estricta de herencia (usando lo que se llama "Mixins").
Caso práctico (Sistema de Archivos):
Tienes una clase Documento y una clase Imagen. Quieres que algunas cosas se puedan exportar a PDF y otras se puedan enviar por correo. Creas clases pequeñas como ExportablePDF y EnviablePorCorreo.
Un ReporteMensual puede heredar de Documento, ExportablePDF y EnviablePorCorreo al mismo tiempo. Obtiene solo las habilidades que necesita, como armar piezas de Lego.

```python
class Documento:
    def __init__(self, contenido):
        self.contenido = contenido

class ExportablePDF:
    def a_pdf(self):
        return "Convirtiendo a PDF..."

class EnviableCorreo:
    def enviar(self, destino):
        return f"Enviando a {destino}..."

class ReporteMensual(Documento, ExportablePDF, EnviableCorreo):
    pass

reporte = ReporteMensual("Datos financieros")
print(reporte.a_pdf())
print(reporte.enviar("jefe@empresa.com"))
```

### 4. Clases Abstractas
El concepto: Es una clase "fantasma" que no se puede usar directamente, sino que sirve como un contrato estricto o plantilla para sus clases hijas.
El problema real que resuelve: En un equipo de trabajo, evita que otro programador olvide implementar una función vital que hará que el sistema explote horas después.
Caso práctico (Conexión a Bases de Datos):
Eres el líder del equipo y creas una clase abstracta ConectorBaseDatos con un método abstracto obligatorio llamado conectar(). Le pides a un programador junior que haga la conexión para MySQL. Si él crea la clase ConectorMySQL pero olvida programar el método conectar(), Python lanzará un error inmediatamente al intentar iniciar el programa, en lugar de fallar en producción cuando un cliente intente guardar sus datos.

```python
from abc import ABC, abstractmethod

class ConectorBaseDatos(ABC):
    @abstractmethod
    def conectar(self):
        pass

class ConectorMySQL(ConectorBaseDatos):
    def conectar(self):
        return "Conexión MySQL establecida"

# bd = ConectorBaseDatos()  # Esto lanzaría TypeError inmediatamente
bd = ConectorMySQL()
print(bd.conectar())
```

### 5. Decoradores
El concepto: Un "envoltorio" que le pones a una función para modificar su comportamiento sin tocar su código interno.
El problema real que resuelve: Ejecutar validaciones o acciones repetitivas antes o después de cientos de funciones distintas.
Caso práctico (Seguridad Web):
Tienes 50 páginas en tu aplicación web. Solo los usuarios con sesión iniciada deberían verlas. En lugar de escribir el código "revisar si hay sesión iniciada" al principio de las 50 funciones, creas un decorador @requiere_login. Solo le pones esa etiqueta encima a las funciones que necesitan protección. El código se mantiene limpio y centralizado.

```python
def requiere_login(funcion):
    def wrapper(usuario):
        if not usuario.get("sesion_iniciada"):
            return "Error: Acceso denegado"
        return funcion(usuario)
    return wrapper

@requiere_login
def ver_panel_control(usuario):
    return "Mostrando datos secretos del panel"

usuario_publico = {"nombre": "Ana", "sesion_iniciada": False}
usuario_admin = {"nombre": "Luis", "sesion_iniciada": True}

print(ver_panel_control(usuario_publico))
print(ver_panel_control(usuario_admin))
```

### 7. Metaclases (Avanzado/Bonus)
El concepto: Clases que fabrican otras clases. Interceptan la creación de una clase antes de que nazca.
El problema real que resuelve: Automatización extrema de frameworks. Raramente lo usarás en el día a día, pero es lo que hace que las librerías que usas parezcan magia.
Caso práctico (Bases de Datos con ORM - Django/SQLAlchemy):
Cuando escribes nombre = Columna(String) dentro de una clase en Python, eso no es código normal. Hay una metaclase oculta en el framework que detecta que escribiste eso, va a la base de datos (PostgreSQL, MySQL) y construye automáticamente una tabla real en el servidor. Tú solo ves código limpio, la metaclase hace el trabajo sucio en segundo plano.

¿Con cuál de estos conceptos sientes que tendrás más dificultades al implementarlo en la lógica de los personajes y habitaciones de tu tarea?

```python
class RegistroORM(type):
    def __new__(cls, nombre, bases, dict_clase):
        nueva_clase = super().__new__(cls, nombre, bases, dict_clase)
        if nombre != "BaseModel":
            print(f"Generando tabla SQL para: {nombre}")
        return nueva_clase

class BaseModel(metaclass=RegistroORM):
    pass

class Usuario(BaseModel):
    pass

class Producto(BaseModel):
    pass
```
